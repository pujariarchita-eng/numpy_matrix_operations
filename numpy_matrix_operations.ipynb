```python
import numpy as np

# Create two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B
print("Matrix Addition:")
print(addition)
```

```python
subtraction = A - B
print("Matrix Subtraction:")
print(subtraction)
```

```python
multiplication = np.dot(A, B)
print("Matrix Multiplication:")
print(multiplication)
```

```python
print("Transpose of Matrix A:")
print(A.T)

print("\nTranspose of Matrix B:")
print(B.T)
```
